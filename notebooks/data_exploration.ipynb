## Dataset split

The exports use the required common keys: `scenario_id`, `simulation_step`, and `simulation_time`. Fault labels are taken from the matching scenario actions rather than included in the deployable measurement matrix. Oracle states are only accepted from explicitly named `oracle` or `hidden` namespaces, which prevents unmarked simulator fields from leaking into measurements.

```python
KEY_COLUMNS = ['scenario_id', 'simulation_step', 'simulation_time']
SOURCE_KEY_COLUMNS = {
    'scenarioId',
    'scenario_id',
    'simulationStep',
    'simulation_step',
    'simulationTime',
    'simulation_time',
}
TRACE_COLUMN_RENAMES = {
    'emulationRealtime': 'emulation_realtime',
    'novaRealtime': 'nova_realtime',
    'runId': 'run_id',
}
PARAMETER_SUFFIXES = frozenset(
    {
        'Acceleration',
        'Deceleration',
        'Length',
        'Mass',
        'MaximumForce',
        'MaximumTorque',
        'PositionLowerLimit',
        'PositionUpperLimit',
        'ProcessTime',
        'Resistance',
        'RollerDamping',
        'Width',
    }
)

if data.empty:
    raise ValueError(f'Cannot split an empty dataset: {INPUT_PATH}')

component_type_knowledge = json.loads(
    (PROJECT_ROOT / 'component_type_knowledge.json').read_text(encoding='utf-8')
)
FAULT_PROPERTIES = frozenset(
    fault['name']
    for component in component_type_knowledge.values()
    for fault in component['faultTypes']
)


def suffix(column: str) -> str:
    return column.rsplit('.', 1)[-1]


def is_oracle_column(column: str) -> bool:
    normalized = column.casefold()
    return normalized.startswith(('oracle.', 'hidden.')) or any(
        marker in normalized for marker in ('.oracle.', '.hidden.')
    )


source_time_column = next(
    (column for column in ('simulationTime', 'simulation_time') if column in data),
    None,
)
if source_time_column is None:
    raise KeyError('The source dataset must contain simulationTime or simulation_time.')

source_step_column = next(
    (column for column in ('simulationStep', 'simulation_step') if column in data),
    None,
)
simulation_steps = (
    data[source_step_column].to_numpy()
    if source_step_column is not None
    else range(len(data))
)
sample_keys = pd.DataFrame(
    {
        'scenario_id': scenario_id,
        'simulation_step': simulation_steps,
        'simulation_time': pd.to_numeric(data[source_time_column]),
    }
)

fault_columns = [
    column for column in data if suffix(column) in FAULT_PROPERTIES
]
oracle_columns = [column for column in data if is_oracle_column(column)]
parameter_candidates = [
    column
    for column in data
    if column not in fault_columns
    and column not in oracle_columns
    and suffix(column) in PARAMETER_SUFFIXES
]
parameter_columns = [
    column
    for column in parameter_candidates
    if data[column].nunique(dropna=False) == 1
]
trace_columns = [column for column in TRACE_COLUMN_RENAMES if column in data]

excluded_columns = (
    SOURCE_KEY_COLUMNS.intersection(data.columns)
    | set(trace_columns)
    | set(fault_columns)
    | set(oracle_columns)
    | set(parameter_columns)
)
measurement_columns = [column for column in data if column not in excluded_columns]

split_groups = {
    'fault': set(fault_columns),
    'oracle': set(oracle_columns),
    'parameter': set(parameter_columns),
    'measurement': set(measurement_columns),
}
for group_name, columns in split_groups.items():
    for other_group_name, other_columns in split_groups.items():
        if group_name < other_group_name and columns.intersection(other_columns):
            raise ValueError(
                f'Columns assigned to both {group_name} and {other_group_name}: '
                f'{sorted(columns.intersection(other_columns))}'
            )

dynamic_parameter_candidates = [
    column for column in parameter_candidates if column not in parameter_columns
]
if dynamic_parameter_candidates:
    print(
        'Keeping dynamic configuration candidates in measurements: '
        + ', '.join(dynamic_parameter_candidates)
    )
```

```python
def clock_to_seconds(clock: str) -> float:
    parts = [float(part) for part in clock.split(':')]
    if len(parts) == 2:
        minutes, seconds = parts
        return minutes * 60 + seconds
    if len(parts) == 3:
        hours, minutes, seconds = parts
        return hours * 3600 + minutes * 60 + seconds
    raise ValueError(f'Unsupported scenario timestamp: {clock}')


def normalize_action_value(value: object) -> str:
    return str(value).strip().replace(',', '.').casefold()


fault_repair_values = {}
for component in component_type_knowledge.values():
    for fault in component['faultTypes']:
        name = fault['name']
        repair_value = normalize_action_value(fault.get('repairValue', ''))
        if name in fault_repair_values and fault_repair_values[name] != repair_value:
            raise ValueError(f'Conflicting repair values defined for {name}.')
        fault_repair_values[name] = repair_value


def empty_fault_events() -> pd.DataFrame:
    return pd.DataFrame(
        {
            'scenario_id': pd.Series(dtype='string'),
            'simulation_step': pd.Series(dtype='int64'),
            'simulation_time': pd.Series(dtype='float64'),
            'fault_area': pd.Series(dtype='string'),
            'fault_component': pd.Series(dtype='string'),
            'fault_property': pd.Series(dtype='string'),
            'fault_value': pd.Series(dtype='string'),
            'event_type': pd.Series(dtype='string'),
            'fault_active': pd.Series(dtype='bool'),
        }
    )


def build_fault_events() -> pd.DataFrame:
    event_rows = []
    for action in scenario['actions']:
        property_name = action.get('property')
        if property_name not in FAULT_PROPERTIES:
            continue

        fault_value = action.get('value', '')
        event_type = (
            'repair'
            if normalize_action_value(fault_value)
            == fault_repair_values[property_name]
            else 'activation'
        )
        event_rows.append(
            {
                'scenario_id': scenario_id,
                'simulation_time': clock_to_seconds(action['startsAt']),
                'fault_area': action.get('area'),
                'fault_component': action.get('element'),
                'fault_property': property_name,
                'fault_value': str(fault_value),
                'event_type': event_type,
                'fault_active': event_type == 'activation',
            }
        )

    if not event_rows:
        return empty_fault_events()

    timeline = sample_keys[['simulation_step', 'simulation_time']].sort_values(
        'simulation_time'
    )
    events = pd.DataFrame(event_rows).sort_values('simulation_time')
    if (
        events['simulation_time'].lt(timeline['simulation_time'].min()).any()
        or events['simulation_time'].gt(timeline['simulation_time'].max()).any()
    ):
        raise ValueError('A fault event falls outside the recorded simulation time range.')

    events = pd.merge_asof(
        events,
        timeline,
        on='simulation_time',
        direction='nearest',
    )
    return events[
        [
            'scenario_id',
            'simulation_step',
            'simulation_time',
            'fault_area',
            'fault_component',
            'fault_property',
            'fault_value',
            'event_type',
            'fault_active',
        ]
    ].reset_index(drop=True)


trace_data = data[trace_columns].rename(columns=TRACE_COLUMN_RENAMES)
measurements = pd.concat(
    [sample_keys, trace_data.reset_index(drop=True), data[measurement_columns]],
    axis=1,
)
oracle_states = pd.concat([sample_keys, data[oracle_columns]], axis=1)
parameters = pd.concat(
    [
        sample_keys.iloc[[0]].reset_index(drop=True),
        data[parameter_columns].iloc[[0]].reset_index(drop=True),
    ],
    axis=1,
)
fault_events = build_fault_events()

if not oracle_columns:
    print(
        'No explicitly named oracle-state columns were found; '
        'oracle_states.parquet contains only the common keys.'
    )

tables = {
    'measurements': measurements,
    'oracle_states': oracle_states,
    'parameters': parameters,
    'fault_events': fault_events,
}
for table_name, table in tables.items():
    missing_keys = set(KEY_COLUMNS).difference(table.columns)
    if missing_keys:
        raise ValueError(f'{table_name} is missing common keys: {sorted(missing_keys)}')

if set(measurement_columns).intersection(fault_columns + oracle_columns):
    raise ValueError('Measurements contain fault or oracle columns.')

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
for table_name, table in tables.items():
    table.to_parquet(OUTPUT_DIR / f'{table_name}.parquet', index=False)

pd.DataFrame(
    {
        'rows': {table_name: len(table) for table_name, table in tables.items()},
        'columns': {
            table_name: len(table.columns) for table_name, table in tables.items()
        },
        'path': {
            table_name: str(OUTPUT_DIR / f'{table_name}.parquet')
            for table_name in tables
        },
    }
)
```
